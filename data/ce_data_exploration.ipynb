## Dataset Structure

**Two-Step Workflow:**
1. **Index Files** (`.npz`): Lightweight files (~0.07 MB/100 samples) that define CE sample structure
2. **Embedding Files** (`.npy`): Large files containing 128-dim fusion embeddings from BEATs + LIMUBert

**Data Splits:**
- **Train**: Subjects 1-4, variable n_samples
- **Val**: Subjects 1-4, fixed n_samples
- **Test (cross-subject)**: Subject 5 - tests generalization to new subject
- **Test (rule-only)**: Subjects 1-4 - tests CE pattern generalization

## 2. Imports and Configuration

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json
import os

sns.set_style('whitegrid')
%matplotlib inline

# Dataset configuration - now supports standard, ae2ce, and softae2ce
CE_BASE_DIR = './CE_dataset/'
DATASET_CONFIG_FILE = './Multimodal/dataset_config_w2.5.json'

# Load label mappings
with open(DATASET_CONFIG_FILE, 'r') as f:
    config = json.load(f)

# AE label mapping (0-8)
ae_label_mapping = config['label_mapping']
ae_labels_inv = {v: k for k, v in ae_label_mapping.items()}

# CE label mapping (0-10)
ce_labels_names = {
    0: 'No violation',
    1: 'Event 1: Post-toilet handwashing',
    2: 'Event 2: Pre-meal handwashing',
    3: 'Event 3: Insufficient brushing (<120s)',
    4: 'Event 4: Post-meal routine',
    5: 'Event 5: Work session pattern',
    6: 'Event 6: Adequate handwashing (30s)',
    7: 'Event 7: Adequate brushing (≥120s)',
    8: 'Event 8: Post-meal rest (180s)',
    9: 'Event 9: Active typing (3 sessions)',
    10: 'Event 10: Focused work (5 clicks)'
}

print("AE Label Mapping (Atomic Activities):")
for name, label in ae_label_mapping.items():
    print(f"  {name:20s}: {label}")

print("\nCE Label Mapping (Complex Events):")
for label, name in ce_labels_names.items():
    print(f"  {label:2d}: {name}")
```

## 3. Dataset Loading Function

```python
def load_ce_dataset(duration=5, split='train', n_samples=None, rule_only=False, dataset_type='standard'):
    """
    Load CE embedding dataset.
    
    Args:
        duration: CE duration in minutes (3, 5, 15, or 30)
        split: Dataset split ('train', 'val', or 'test')
        n_samples: Number of samples (None to auto-detect, required for train)
        rule_only: Load rule-only test variant (only for split='test')
        dataset_type: Dataset type ('standard', 'ae2ce', 'softae2ce', 'shifted', 'nar')
    
    Returns:
        tuple: (embeddings, ce_labels, info)
    """
    # Determine subdirectory and filename prefix based on dataset type
    if dataset_type in ['standard', 'shifted']:
        data_dir = Path(CE_BASE_DIR) / dataset_type
        prefix = f'ce{duration}min'
    elif dataset_type == 'ae2ce':
        data_dir = Path(CE_BASE_DIR) / 'ae2ce'
        prefix = f'ae2ce{duration}min'
    elif dataset_type == 'softae2ce':
        data_dir = Path(CE_BASE_DIR) / 'softae2ce'
        prefix = f'softae2ce{duration}min'
    elif dataset_type == 'nar':
        data_dir = Path(CE_BASE_DIR) / 'nar'
        prefix = f'ct_ce{duration}min'
    else:
        raise ValueError(f"Invalid dataset_type: {dataset_type}")
    
    # Construct filenames
    if split == 'train':
        if n_samples is None:
            # Try to find any train file
            pattern = f'{prefix}_train_*_data*.npy'
            matches = list(data_dir.glob(pattern))
            if not matches:
                raise FileNotFoundError(f"No CE dataset found for duration={duration}, split={split}, type={dataset_type}")
            data_file = matches[0]
            # Fix: replace '_data_' not '_data.npy' to handle '_data_2000.npy' correctly
            labels_file = Path(str(data_file).replace('_data_', '_labels_'))
        else:
            # Fix: construct labels filename explicitly
            data_file = data_dir / f'{prefix}_train_data_{n_samples}.npy'
            labels_file = data_dir / f'{prefix}_train_labels_{n_samples}.npy'
    elif split == 'val':
        data_file = data_dir / f'{prefix}_val_data.npy'
        labels_file = data_dir / f'{prefix}_val_labels.npy'
    elif split == 'test':
        if rule_only:
            data_file = data_dir / f'{prefix}_test_rule_data.npy'
            labels_file = data_dir / f'{prefix}_test_rule_labels.npy'
        else:
            data_file = data_dir / f'{prefix}_test_data.npy'
            labels_file = data_dir / f'{prefix}_test_labels.npy'
    else:
        raise ValueError(f"Invalid split: {split}")
    
    if not data_file.exists():
        raise FileNotFoundError(f"Data file not found: {data_file}")
    if not labels_file.exists():
        raise FileNotFoundError(f"Labels file not found: {labels_file}")
    
    # Load data
    embeddings = np.load(data_file)
    ce_labels = np.load(labels_file)
    
    info = {
        'duration': duration,
        'split': split,
        'rule_only': rule_only,
        'dataset_type': dataset_type,
        'n_samples': embeddings.shape[0],
        'n_windows': embeddings.shape[1],
        'embed_dim': embeddings.shape[2] if embeddings.ndim == 3 else None,
        'data_file': str(data_file),
        'labels_file': str(labels_file)
    }
    
    return embeddings, ce_labels, info
```

## 5. Load Dataset

Modify the parameters below to load different datasets.

```python
# Load a dataset (modify parameters as needed)
duration = 5
split = 'train'
n_samples = 2000  # Required for train split
dataset_type = 'standard'  # Options: 'standard', 'ae2ce', 'softae2ce', 'shifted', 'nar'

embeddings, ce_labels, info = load_ce_dataset(
    duration=duration, 
    split=split, 
    n_samples=n_samples,
    dataset_type=dataset_type
)

print("="*70)
print("Dataset Loaded Successfully")
print("="*70)
for key, value in info.items():
    print(f"  {key:15s}: {value}")

print(f"\nEmbeddings Shape: {embeddings.shape}  (samples, windows, features)")
print(f"CE Labels Shape:  {ce_labels.shape}  (samples, windows, labels)")

# Show what the data represents based on dataset type
if dataset_type in ['ae2ce', 'softae2ce']:
    print(f"\nNote: This is an {dataset_type.upper()} dataset")
    print(f"  - Feature dimension: {embeddings.shape[2]} (9 atomic activity classes)")
    if dataset_type == 'ae2ce':
        print(f"  - Data format: One-hot encoded predictions (binary 0/1)")
    else:
        print(f"  - Data format: Soft probability distributions (sum=1.0)")
```

## 6. View Sample Data

```python
# Analyze multi-label samples
print("="*70)
print("Multi-Label Analysis")
print("="*70)

# Count how many events are active per window across all samples
events_per_window = np.sum(ce_labels, axis=2)  # Sum across event dimension -> (n_samples, n_windows)

# Statistics
print(f"\nEvents per window (across all samples):")
print(f"  Mean:   {events_per_window.mean():.3f}")
print(f"  Median: {np.median(events_per_window):.1f}")
print(f"  Max:    {int(events_per_window.max())}")

# Count windows with multiple events
multi_event_windows = np.sum(events_per_window > 1)
total_windows = events_per_window.size
print(f"\nWindows with multiple events: {multi_event_windows}/{total_windows} ({multi_event_windows/total_windows*100:.2f}%)")

# Count samples with multi-label windows
samples_with_multilabel = np.sum(np.any(events_per_window > 1, axis=1))
n_samples = ce_labels.shape[0]
print(f"Samples containing multi-label windows: {samples_with_multilabel}/{n_samples} ({samples_with_multilabel/n_samples*100:.2f}%)")

# Distribution of number of simultaneous events (window-level)
print(f"\n" + "-"*70)
print("Window-Level Distribution:")
print("-"*70)
print(f"\nDistribution of simultaneous events per window:")
unique_counts, count_freqs = np.unique(events_per_window, return_counts=True)
for n_events, freq in zip(unique_counts, count_freqs):
    print(f"  {int(n_events)} events: {freq:6d} windows ({freq/total_windows*100:5.2f}%)")

# Sample-level distribution
print(f"\n" + "-"*70)
print("Sample-Level Distribution:")
print("-"*70)

# Check if each sample has at least one event
has_any_event = np.any(events_per_window > 0, axis=1)  # (n_samples,)
samples_with_events = np.sum(has_any_event)
samples_without_events = n_samples - samples_with_events

print(f"\nSamples by event presence:")
print(f"  Samples with 0 events (all windows = 0):      {samples_without_events:4d} ({samples_without_events/n_samples*100:5.2f}%)")
print(f"  Samples with at least 1 event (any window > 0): {samples_with_events:4d} ({samples_with_events/n_samples*100:5.2f}%)")

# Show distribution of total events per sample
total_events_per_sample = np.sum(events_per_window, axis=1)  # Sum across all windows
print(f"\nTotal event occurrences per sample (across all 120 windows):")
print(f"  Mean:   {total_events_per_sample.mean():.2f}")
print(f"  Median: {np.median(total_events_per_sample):.0f}")
print(f"  Min:    {int(total_events_per_sample.min())}")
print(f"  Max:    {int(total_events_per_sample.max())}")

# Analyze event co-occurrence patterns
print(f"\n" + "="*70)
print("Event Co-occurrence Analysis")
print("="*70)

# Build co-occurrence matrix
n_events = 10
cooccurrence_matrix = np.zeros((n_events, n_events), dtype=int)

# Reshape to (total_windows, n_events) for easier processing
labels_flat = ce_labels.reshape(-1, n_events)  # (n_samples*n_windows, 10)

for window_labels in labels_flat:
    active_events = np.where(window_labels == 1)[0]  # Get indices of active events
    # Count co-occurrences for all pairs
    for i in active_events:
        for j in active_events:
            if i != j:
                cooccurrence_matrix[i, j] += 1

# Show most common event pairs
print("\nMost common event pairs (co-occurring in same window):")
pair_counts = []
for i in range(n_events):
    for j in range(i+1, n_events):  # Only upper triangle to avoid duplicates
        count = cooccurrence_matrix[i, j] + cooccurrence_matrix[j, i]
        if count > 0:
            pair_counts.append((count, i+1, j+1))  # +1 for event IDs (1-10)

pair_counts.sort(reverse=True)
if len(pair_counts) > 0:
    for count, ev1, ev2 in pair_counts[:15]:  # Top 15 pairs
        print(f"  Event {ev1} + Event {ev2}: {count:4d} windows ({count/total_windows*100:5.2f}%)")
        print(f"    - {ce_labels_names[ev1]}")
        print(f"    - {ce_labels_names[ev2]}")
else:
    print("  No event co-occurrences found in this dataset")

# Find samples with most multi-label activity
max_events_per_sample = np.max(events_per_window, axis=1)
top_multilabel_samples = np.argsort(max_events_per_sample)[-5:][::-1]

print(f"\n" + "="*70)
print("Top Samples with Multi-Label Windows")
print("="*70)
for rank, idx in enumerate(top_multilabel_samples, 1):
    max_sim = int(max_events_per_sample[idx])
    n_multilabel_windows = int(np.sum(events_per_window[idx] > 1))
    
    if max_sim > 1:  # Only show samples that actually have multi-label windows
        print(f"\n{rank}. Sample {idx}: Max {max_sim} simultaneous events, {n_multilabel_windows} multi-label windows")
        
        # Show which events co-occur in this sample
        sample_labels = ce_labels[idx]  # (120, 10)
        for w in range(120):
            active_events = np.where(sample_labels[w] == 1)[0] + 1  # +1 for event IDs
            if len(active_events) > 1:
                event_names = [f"E{e}" for e in active_events]
                print(f"  Window {w:3d}: {', '.join(event_names)}")
```

def print_dataset_statistics(embeddings, ce_labels):
    """
    Print comprehensive dataset statistics.
    """
    print("=" * 70)
    print("Dataset Statistics")
    print("=" * 70)
    
    # Embedding statistics
    print("\nEmbedding Statistics (128-dim fusion vectors):")
    print(f"  Shape: {embeddings.shape}")
    print(f"  Mean across all: {embeddings.mean():.4f}")
    print(f"  Std across all:  {embeddings.std():.4f}")
    print(f"  Min value:       {embeddings.min():.4f}")
    print(f"  Max value:       {embeddings.max():.4f}")
    
    # CE label distribution
    print("\nComplex Event Label Distribution:")
    unique_ce, counts_ce = np.unique(ce_labels, return_counts=True)
    total_windows = ce_labels.size
    for label, count in zip(unique_ce, counts_ce):
        ce_name = ce_labels_names.get(int(label), f'Unknown {label}')
        print(f"  {int(label):2d}: {int(count):6d} windows ({count/total_windows*100:5.2f}%) - {ce_name}")
    
    # CE violation statistics

    print(f"  Mean:   {violations_per_sample.mean():.2f} windows")

    print(f"  Min:    {int(violations_per_sample.min())} windows")
    print(f"  Max:    {int(violations_per_sample.max())} windows")
